# 第5章：LLM基础 - 交互式学习

本notebook提供LLM基础概念的交互式学习体验。

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Scaling Laws演示
model_sizes = np.array([1, 10, 100, 1000, 10000])  # 百万参数
data_sizes = np.array([1, 10, 100, 1000, 10000])   # 百万tokens

# 根据Chinchilla scaling laws
loss_model = 1.0 / (model_sizes ** 0.07)
loss_data = 1.0 / (data_sizes ** 0.09)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 模型大小 vs 性能
axes[0].loglog(model_sizes, loss_model, 'o-', linewidth=2, markersize=8)
axes[0].set_xlabel('Model Size (Millions of Parameters)')
axes[0].set_ylabel('Loss')
axes[0].set_title('Model Size vs Performance')
axes[0].grid(True, alpha=0.3)

# 数据大小 vs 性能
axes[1].loglog(data_sizes, loss_data, 's-', linewidth=2, markersize=8, color='orange')
axes[1].set_xlabel('Data Size (Millions of Tokens)')
axes[1].set_ylabel('Loss')
axes[1].set_title('Data Size vs Performance')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Scaling Laws: 性能随模型和数据大小的增加而改进")
```
