```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# RAG系统演示
print("RAG (Retrieval-Augmented Generation) 系统流程:")
print("="*50)
print("1. 用户提问")
print("2. 检索相关文档")
print("3. 将文档和问题传给LLM")
print("4. LLM生成答案")
print("="*50)

# 模拟检索结果
query = "什么是Transformer?"
documents = [
    "Transformer是一种基于自注意力机制的神经网络架构...",
    "Transformer由编码器和解码器组成...",
    "自注意力机制允许模型关注输入的不同部分..."
]

print(f"\n查询: {query}")
print(f"\n检索到 {len(documents)} 个相关文档:")
for i, doc in enumerate(documents, 1):
    print(f"  {i}. {doc[:50]}...")
```
